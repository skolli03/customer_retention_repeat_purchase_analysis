```python
import pandas as pd

# 1. Load orders and pick 15% of users
orders = pd.read_csv("orders.csv")
users = orders["user_id"].drop_duplicates().sample(frac=0.15, random_state=42)
orders_s = orders[orders["user_id"].isin(users)]

# 2. Keep only prior orders (these have product data)
orders_s = orders_s[orders_s["eval_set"] == "prior"]
orders_s.to_csv("orders_sample.csv", index=False)

# 3. Keep only products from those orders, reading in chunks
keep_ids = set(orders_s["order_id"])
first = True
for chunk in pd.read_csv("order_products__prior.csv", chunksize=1_000_000):
    out = chunk[chunk["order_id"].isin(keep_ids)]
    out.to_csv("order_products_sample.csv", mode="w" if first else "a",
               header=first, index=False)
    first = False

print("Done:", len(orders_s), "orders in sample")
```
